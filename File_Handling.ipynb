```python
import csv
import os

FILE_NAME = "expenses.csv"

# Create the CSV file with headings only if it doesn't exist yet
if not os.path.exists(FILE_NAME):
    with open(FILE_NAME, "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["Category", "Amount"])


def add_expense():
    category = input("Enter expense category (e.g. Food, Travel): ").strip()
    amount_text = input("Enter amount spent: ").strip()

    if category == "" or amount_text == "":
        print("Category and amount cannot be empty.")
        return

    try:
        amount = float(amount_text)
        if amount <= 0:
            print("Amount must be greater than zero.")
            return
    except ValueError:
        print("Please enter a valid number for amount.")
        return

    with open(FILE_NAME, "a", newline="") as file:
        writer = csv.writer(file)
        writer.writerow([category, amount])

    print("Expense saved successfully.")


def view_expenses():
    try:
        with open(FILE_NAME, "r") as file:
            reader = csv.reader(file)
            print("\n===== ALL EXPENSES =====")
            for row in reader:
                print(row)
    except FileNotFoundError:
        print("No expenses recorded yet.")


def show_total():
    total = 0
    try:
        with open(FILE_NAME, "r") as file:
            reader = csv.reader(file)
            next(reader)  # skip header row
            for row in reader:
                try:
                    total += float(row[1])
                except (ValueError, IndexError):
                    continue  # skip any bad/corrupted rows
        print(f"\nTotal spent so far: {total}")
    except FileNotFoundError:
        print("No expenses recorded yet.")


def main():
    while True:
        print("\n===== EXPENSE TRACKER =====")
        print("1. Add Expense")
        print("2. View Expenses")
        print("3. Show Total Spent")
        print("4. Exit")
        choice = input("Enter your choice: ")

        if choice == "1":
            add_expense()
        elif choice == "2":
            view_expenses()
        elif choice == "3":
            show_total()
        elif choice == "4":
            print("Expense tracker closed.")
            break
        else:
            print("Invalid choice. Please select 1, 2, 3, or 4.")


if __name__ == "__main__":
    main()1
```
